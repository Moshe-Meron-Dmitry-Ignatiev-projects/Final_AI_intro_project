```python
# 1

import fitz  # PyMuPDF
import easyocr
import os
import json
from PIL import Image
import io
from tqdm import tqdm  # בר התקדמות

# אתחול OCR
reader = easyocr.Reader(['en'])

# יצירת תיקייה לתמונות
os.makedirs("extracted_images", exist_ok=True)

# פתיחת הקובץ
pdf_path = "US11960514.pdf"
doc = fitz.open(pdf_path)

# רשימת מטא־דאטה
metadata = []

# פונקציית סיווג טקסט
def classify_text(text):
    text_lower = text.lower()
    if text_lower.startswith("abstract"):
        return "abstract"
    elif text_lower.startswith("claim") or "claims" in text_lower:
        return "claims"
    elif text_lower.startswith("title") or len(text.split()) < 5:
        return "title"
    elif "table" in text_lower:
        return "table"
    elif any(sym in text_lower for sym in ["=", "+", "-", "∑", "∫"]):
        return "formula"
    elif "figure" in text_lower or "fig." in text_lower:
        return "caption"
    else:
        return "text"

# עיבוד עם בר התקדמות
for page_number in tqdm(range(len(doc)), desc="Processing PDF", unit="page"):
    page = doc.load_page(page_number)
    pix = page.get_pixmap(dpi=300)
    img_bytes = pix.tobytes("png")
    image = Image.open(io.BytesIO(img_bytes))

    # OCR
    ocr_result = reader.readtext(img_bytes, detail=0)
    extracted_text = "\n".join(ocr_result).strip()

    if extracted_text:
        classification = classify_text(extracted_text)
        metadata.append({
            "type": classification,
            "page": page_number + 1,
            "content": extracted_text
        })

    # חילוץ תמונות
    image_list = page.get_images(full=True)
    for img_index, img in enumerate(image_list):
        xref = img[0]
        base_image = doc.extract_image(xref)
        image_bytes = base_image["image"]
        image_ext = base_image["ext"]
        image_filename = f"extracted_images/page_{page_number + 1}_image_{img_index + 1}.{image_ext}"
        with open(image_filename, "wb") as img_file:
            img_file.write(image_bytes)
        metadata.append({
            "type": "image",
            "page": page_number + 1,
            "content": image_filename
        })

# שמירת המטא־דאטה
with open("metadata.json", "w", encoding="utf-8") as f:
    json.dump(metadata, f, ensure_ascii=False, indent=2)

print("✅ סיום! כל הדפים עובדו. המטא־דאטה נשמרה בקובץ metadata.json.")
print(f"Processed {len(doc)} pages. Metadata saved to metadata.json.")
```

```python
# 2

import json
import numpy as np
from sentence_transformers import SentenceTransformer
from qdrant_client import QdrantClient
from qdrant_client.http.models import Distance, VectorParams, PointStruct

# טען את המטא־דאטה
with open("metadata.json", "r", encoding="utf-8") as f:
    metadata = json.load(f)

# אתחול מודל הטמעה
model = SentenceTransformer("all-MiniLM-L6-v2")

# אתחול לקוח Qdrant בזיכרון
qdrant = QdrantClient(":memory:")

# יצירת אוסף חדש
collection_name = "patent_chunks"
qdrant.recreate_collection(
    collection_name=collection_name,
    vectors_config=VectorParams(size=384, distance=Distance.COSINE)
)

# המרת טקסטים לווקטורים והוספה ל־Qdrant
points = []
for idx, item in enumerate(metadata):
    if item["type"] != "image":
        vector = model.encode(item["content"]).tolist()
        points.append(PointStruct(
            id=idx,
            vector=vector,
            payload={
                "type": item["type"],
                "page": item["page"],
                "content": item["content"]
            }
        ))

# אחסון הווקטורים
qdrant.upsert(collection_name=collection_name, points=points)

print(f"✅ {len(points)} וקטורים הוטמעו ונשמרו ב־Qdrant.")

# סגירת Qdrant (אם נדרש לשחרר משאבים)
qdrant.close()
print("🛑 Qdrant נסגר. ניתן לגשת אליו מחדש בשלבים הבאים.")
```

```python
# 4

import json
from sentence_transformers import SentenceTransformer
from qdrant_client import QdrantClient
from qdrant_client.http.models import Distance, VectorParams, PointStruct

# הגדרות
MAX_CONTEXT_BYTES = 2000
TOP_K = 5

# טען שאלות
with open("questions.txt", "r", encoding="utf-8") as f:
    questions = [line.strip() for line in f if line.strip()]

# טען מטא־דאטה
with open("metadata.json", "r", encoding="utf-8") as f:
    metadata = json.load(f)

# אתחול מודל הטמעה
model = SentenceTransformer("all-MiniLM-L6-v2")

# אתחול Qdrant בזיכרון
qdrant = QdrantClient(":memory:")
collection_name = "patent_chunks"

# יצירת אוסף מחדש
qdrant.recreate_collection(
    collection_name=collection_name,
    vectors_config=VectorParams(size=384, distance=Distance.COSINE)
)

# הטמעת טקסטים והוספה ל־Qdrant
points = []
for idx, item in enumerate(metadata):
    if item["type"] != "image":
        vector = model.encode(item["content"]).tolist()
        points.append(PointStruct(
            id=idx,
            vector=vector,
            payload={
                "type": item["type"],
                "page": item["page"],
                "content": item["content"]
            }
        ))
qdrant.upsert(collection_name=collection_name, points=points)

# בניית פרומפטים
prompts = []
for question in questions:
    query_vector = model.encode(question).tolist()
    search_result = qdrant.search(collection_name=collection_name, query_vector=query_vector, limit=TOP_K)

    selected_chunks = []
    pages = set()
    total_bytes = 0
    for hit in search_result:
        chunk_text = hit.payload["content"]
        chunk_bytes = len(chunk_text.encode("utf-8"))
        if total_bytes + chunk_bytes <= MAX_CONTEXT_BYTES:
            selected_chunks.append(chunk_text)
            total_bytes += chunk_bytes
            pages.add(hit.payload["page"])
        else:
            break

    # שליפת תמונות סמוכות
    image_paths = []
    image_pages = set()
    for item in metadata:
        if item["type"] == "image" and item["page"] in pages:
            image_paths.append(item["content"])
            image_pages.add(item["page"])
            if len(image_paths) == 2:
                break

    # בניית הפרומפט עם הנחיה ברורה ותיעוד עמודים
    instruction = "Use only the information provided in the patent document below. Do not rely on external knowledge."
    context_text = "\n---\n".join(selected_chunks)
    image_text = ", ".join(image_paths)
    page_info = f"Pages used: {sorted(pages.union(image_pages))}"
    prompt = f"{instruction}\n\nQuestion: {question}\n\nContext:\n{context_text}\n\nImages:\n{image_text}\n\n{page_info}"
    prompts.append(prompt)

# שמירה לקובץ
with open("prompts.txt", "w", encoding="utf-8") as f:
    for p in prompts:
        f.write(p + "\n\n")

# סגירת Qdrant
qdrant.close()
print("✅ כל הפרומפטים נבנו ונשמרו בקובץ prompts.txt עם תיעוד של מספרי העמודים.")

```

```python
# שלב 5 נסיוני 

import os
import subprocess
from tqdm import tqdm

# יצירת קובץ תשובות אם אינו קיים
answers_file = "answers.txt"
if not os.path.exists(answers_file):
    with open(answers_file, "w", encoding="utf-8") as f:
        f.write("")

# פונקציה להרצת LLaVA על תמונה
def run_llava(image_path):
    command = ["ollama", "run", "llava", f"<image:{image_path}> Describe the image."]
    result = subprocess.run(command, capture_output=True, text=True)
    return result.stdout.strip()

# פונקציה להרצת LLaMA על פרומפט טקסטואלי
def run_llama(prompt):
    command = ["ollama", "run", "llama3", prompt]
    result = subprocess.run(command, capture_output=True, text=True)
    return result.stdout.strip()[:300]

# קריאת הפרומפטים
with open("prompts.txt", "r", encoding="utf-8") as f:
    raw_prompts = f.read().strip().split("\n\n")

answers = []

# בר התקדמות
for prompt_block in tqdm(raw_prompts, desc="Generating answers", unit="question"):
    lines = prompt_block.split("\n")
    question_line = next((line for line in lines if line.startswith("Question:")), "")
    image_line = next((line for line in lines if line.startswith("Images:")), "")
    
    # פענוח תמונות אם קיימות
    image_descriptions = []
    image_paths = image_line.replace("Images:", "").strip().split(", ")
    for image_path in image_paths:
        if image_path and os.path.exists(image_path):
            description = run_llava(image_path)
            image_descriptions.append(description)
    
    # הוספת תיאורי תמונות להקשר
    enriched_prompt = prompt_block
    if image_descriptions:
        enriched_prompt += "\n\nImage Descriptions:\n" + "\n".join(image_descriptions)
    
    # הפקת תשובה עם LLaMA
    answer = run_llama(enriched_prompt)
    answers.append(f"{question_line}\nAnswer: {answer}")

# כתיבה לקובץ התשובות
with open(answers_file, "a", encoding="utf-8") as f:
    for a in answers:
        f.write(a + "\n\n")

print("✅ כל התשובות נוצרו ונשמרו בקובץ answers.txt.")

```

```python
# שלב 5 עם טיפול בבעית קידוד - לא נוסה

import os
import subprocess
from tqdm import tqdm

answers_file = "answers.txt"

# יצירת קובץ תשובות אם אינו קיים
if not os.path.exists(answers_file):
    with open(answers_file, "w", encoding="utf-8") as f:
        f.write("")

# פונקציה להרצת LLaVA על תמונה
def run_llava(image_path):
    try:
        command = ["ollama", "run", "llava", f"<image:{image_path}> Describe the image."]
        result = subprocess.run(command, capture_output=True, text=True, encoding="utf-8")
        return result.stdout.strip()
    except Exception as e:
        return f"[LLaVA error: {e}]"

# פונקציה להרצת LLaMA על פרומפט טקסטואלי
def run_llama(prompt):
    try:
        command = ["ollama", "run", "llama3", prompt]
        result = subprocess.run(command, capture_output=True, text=True, encoding="utf-8")
        return result.stdout.strip()[:300]
    except Exception as e:
        return f"[LLaMA error: {e}]"

# קריאת הפרומפטים
with open("prompts.txt", "r", encoding="utf-8") as f:
    raw_prompts = f.read().strip().split("\n\n")

answers = []

# בר התקדמות
for prompt_block in tqdm(raw_prompts, desc="Generating answers", unit="question"):
    lines = prompt_block.split("\n")
    question_line = next((line for line in lines if line.startswith("Question:")), "")
    image_line = next((line for line in lines if line.startswith("Images:")), "")
    
    # פענוח תמונות אם קיימות
    image_descriptions = []
    image_paths = image_line.replace("Images:", "").strip().split(", ")
    for image_path in image_paths:
        image_path = image_path.strip()
        if image_path and os.path.exists(image_path):
            description = run_llava(image_path)
            image_descriptions.append(description)
    
    # הוספת תיאורי תמונות להקשר
    enriched_prompt = prompt_block
    if image_descriptions:
        enriched_prompt += "\n\nImage Descriptions:\n" + "\n".join(image_descriptions)
    
    # הפקת תשובה עם LLaMA
    answer = run_llama(enriched_prompt)
    answers.append(f"{question_line}\nAnswer: {answer}")

# כתיבה לקובץ התשובות
with open(answers_file, "a", encoding="utf-8") as f:
    for a in answers:
        f.write(a + "\n\n")

print("✅ כל התשובות נוצרו ונשמרו בקובץ answers.txt.")
```

```python
# 5

import os
import subprocess

def run_llava(image_path):
    """הרצת LLaVA על תמונה אחת והחזרת תיאור טקסטואלי"""
    command = ["ollama", "run", "llava", f"<image:{image_path}> Describe the image."]
    result = subprocess.run(command, capture_output=True, text=True)
    return result.stdout.strip()

def run_llama(prompt):
    """הרצת LLaMA על פרומפט טקסטואלי והחזרת תשובה מוגבלת ל־300 תווים"""
    command = ["ollama", "run", "llama3", prompt]
    result = subprocess.run(command, capture_output=True, text=True)
    return result.stdout.strip()[:300]

# קריאת הפרומפטים
with open("prompts.txt", "r", encoding="utf-8") as f:
    raw_prompts = f.read().strip().split("\n\n")

answers = []

for prompt_block in raw_prompts:
    lines = prompt_block.split("\n")
    question_line = next((line for line in lines if line.startswith("Question:")), "")
    image_line = next((line for line in lines if line.startswith("Images:")), "")
    
    # פענוח תמונות אם קיימות
    image_descriptions = []
    image_paths = image_line.replace("Images:", "").strip().split(", ")
    for image_path in image_paths:
        if image_path and os.path.exists(image_path):
            description = run_llava(image_path)
            image_descriptions.append(description)
    
    # הוספת תיאורי תמונות להקשר
    enriched_prompt = prompt_block
    if image_descriptions:
        enriched_prompt += "\n\nImage Descriptions:\n" + "\n".join(image_descriptions)
    
    # הפקת תשובה עם LLaMA
    answer = run_llama(enriched_prompt)
    answers.append(f"{question_line}\nAnswer: {answer}")

# שמירת התשובות
with open("answers.txt", "w", encoding="utf-8") as f:
    for a in answers:
        f.write(a + "\n\n")

print("✅ תשובות נוצרו ונשמרו בקובץ answers.txt.")
```
