```python
import os
import fitz  # PyMuPDF
import json
import numpy as np
import easyocr
from sentence_transformers import SentenceTransformer
from qdrant_client import QdrantClient
from qdrant_client.http.models import VectorParams, PointStruct

# ----------------------------------------
# Configuration
# ----------------------------------------
PDF_PATH               = "patent.pdf"
IMAGES_DIR             = "images"
VECTOR_MODEL_NAME      = "all-MiniLM-L6-v2"
QDRANT_COLLECTION_NAME = "patent_chunks"

# ----------------------------------------
# Prepare folders & OCR reader
# ----------------------------------------
os.makedirs(IMAGES_DIR, exist_ok=True)
ocr_reader = easyocr.Reader(["en"])  # OCR for English text

# ----------------------------------------
# Helper: Check for duplicate chunks
# ----------------------------------------
def is_duplicate(new_chunk, existing_chunks):
    for chunk in existing_chunks:
        if chunk["page"] == new_chunk["page"] and chunk["type"] == new_chunk["type"]:
            if chunk["content"].strip() == new_chunk["content"].strip():
                return True
    return False

# ----------------------------------------
# Step 1: Chunking the Patent
# ----------------------------------------
chunks = []
doc = fitz.open(PDF_PATH)

for page_index in range(len(doc)):
    page_number = page_index + 1
    page = doc.load_page(page_index)

    # 1. Extract embedded text
    embedded_text = page.get_text().strip()

    # 2. Extract inline images
    images = page.get_images(full=True)

    # 3a. Prefer embedded text if available
    if embedded_text and not is_duplicate({"type": "text", "page": page_number, "content": embedded_text}, chunks):
        chunks.append({
            "type": "text",
            "page": page_number,
            "content": embedded_text
        })

    # 3b. Extract and save images only if they are diagrams or full images
    for img_idx, img in enumerate(images, start=1):
        xref     = img[0]
        img_dict = doc.extract_image(xref)
        img_bytes= img_dict["image"]
        ext      = img_dict["ext"]
        fname    = f"page_{page_number}_img_{img_idx}.{ext}"
        fpath    = os.path.join(IMAGES_DIR, fname)
        with open(fpath, "wb") as f:
            f.write(img_bytes)

        # Only add image chunk if no embedded text OR image is clearly schematic
        if not embedded_text:
            chunks.append({
                "type": "image",
                "page": page_number,
                "content": fpath
            })

    # 3c. If no embedded text and no inline images, run OCR on full page
    if not embedded_text and not images:
        pix = page.get_pixmap()
        img_bytes = pix.tobytes()
        ocr_result = ocr_reader.readtext(np.frombuffer(img_bytes, dtype=np.uint8))
        ocr_text = " ".join([line[1] for line in ocr_result]).strip()
        if ocr_text and not is_duplicate({"type": "text", "page": page_number, "content": ocr_text}, chunks):
            chunks.append({
                "type": "text",
                "page": page_number,
                "content": ocr_text
            })

# Optional: save chunks metadata for inspection
with open("chunks.json", "w", encoding="utf-8") as jf:
    json.dump(chunks, jf, ensure_ascii=False, indent=2)

# ----------------------------------------
# Step 2: Vector Store
# ----------------------------------------

# 1. Filter only text chunks
text_chunks = [c["content"] for c in chunks if c["type"] == "text"]

if text_chunks:
    # 2. Encode to vectors
    encoder = SentenceTransformer(VECTOR_MODEL_NAME)
    vectors = encoder.encode(text_chunks, show_progress_bar=True)

    # 3. Initialize in-memory Qdrant
    client = QdrantClient(":memory:")
    dim = vectors.shape[1]
    client.create_collection(
        collection_name=QDRANT_COLLECTION_NAME,
        vectors_config=VectorParams(size=dim, distance="Cosine")
    )

    # 4. Upsert points (id, vector, payload)
    points = [
        PointStruct(
            id=i,
            vector=vectors[i].tolist(),
            payload={
                "page": chunks[i]["page"],
                "content": text_chunks[i]
            }
        )
        for i in range(len(vectors))
    ]
    client.upsert(collection_name=QDRANT_COLLECTION_NAME, points=points)

    print(f"🗄️ Upserted {len(points)} text chunks into Qdrant collection '{QDRANT_COLLECTION_NAME}'.")
else:
    print("⚠️ No text chunks found for vectorization.")
```

```python
import os
import json
from qdrant_client import QdrantClient
from sentence_transformers import SentenceTransformer

# ----------------------------------------
# Configuration
# ----------------------------------------
QUESTIONS_FILE         = "questions.txt"
CHUNKS_FILE            = "chunks.json"
MAX_CONTEXT_BYTES      = 2000
TOP_K_TEXT_CHUNKS      = 5
MAX_IMAGES_PER_PROMPT  = 2
QDRANT_COLLECTION_NAME = "patent_chunks"
VECTOR_MODEL_NAME      = "all-MiniLM-L6-v2"

# ----------------------------------------
# Load questions
# ----------------------------------------
with open(QUESTIONS_FILE, "r", encoding="utf-8") as f:
    questions = [line.strip() for line in f if line.strip()]

# ----------------------------------------
# Load chunks metadata
# ----------------------------------------
with open(CHUNKS_FILE, "r", encoding="utf-8") as jf:
    all_chunks = json.load(jf)

# Separate chunks by type
text_chunks = [c for c in all_chunks if c["type"] == "text"]
image_chunks = [c for c in all_chunks if c["type"] == "image"]

# ----------------------------------------
# Initialize encoder and Qdrant
# ----------------------------------------
encoder = SentenceTransformer(VECTOR_MODEL_NAME)
client = QdrantClient(":memory:")

# Rebuild Qdrant collection
vectors = encoder.encode([c["content"] for c in text_chunks])
dim = vectors.shape[1]
client.create_collection(
    collection_name=QDRANT_COLLECTION_NAME,
    vectors_config={"size": dim, "distance": "Cosine"}
)

from qdrant_client.http.models import PointStruct

points = [
    PointStruct(
        id=i,
        vector=vectors[i].tolist(),
        payload={
            "page": text_chunks[i]["page"],
            "content": text_chunks[i]["content"]
        }
    )
    for i in range(len(text_chunks))
]

client.upsert(collection_name=QDRANT_COLLECTION_NAME, points=points)

# ----------------------------------------
# Step 4: Construct prompts
# ----------------------------------------
prompts = []

for q_index, question in enumerate(questions):
    print(f"\n🔍 Processing Question {q_index+1}: {question}")

    # Encode question
    q_vector = encoder.encode([question])[0].tolist()

    # Search top-k relevant text chunks
    search_results = client.search(
        collection_name=QDRANT_COLLECTION_NAME,
        query_vector=q_vector,
        limit=TOP_K_TEXT_CHUNKS
    )

    # Collect context text
    context_chunks = []
    total_bytes = 0
    for result in search_results:
        text = result.payload["content"]
        text_bytes = len(text.encode("utf-8"))
        if total_bytes + text_bytes <= MAX_CONTEXT_BYTES:
            context_chunks.append(text)
            total_bytes += text_bytes
        else:
            break

    # Collect image paths from same/adjacent pages
    relevant_pages = set()
    for result in search_results:
        page = result.payload["page"]
        relevant_pages.update([page - 1, page, page + 1])

    selected_images = []
    for img in image_chunks:
        if img["page"] in relevant_pages:
            selected_images.append(img["content"])
        if len(selected_images) >= MAX_IMAGES_PER_PROMPT:
            break

    # Build prompt safely
    context_text = "\n".join(context_chunks)
    image_text = ", ".join(selected_images)

    prompt = f"""Question: {question}

Context:
{context_text}

Images:
{image_text}
"""
    prompts.append(prompt)
    print(prompt)

# Optional: save prompts to file
with open("prompts.txt", "w", encoding="utf-8") as pf:
    for p in prompts:
        pf.write(p + "\n" + "="*80 + "\n")
```

```python
import os
import json
import subprocess

# ----------------------------------------
# Configuration
# ----------------------------------------
PROMPTS_FILE       = "prompts.txt"
LLAVA_MODEL        = "llava"
LLAMA_MODEL        = "llama3.2:1b"
MAX_ANSWER_CHARS   = 300
MAX_IMAGE_TOKENS   = 50

SYSTEM_LLAVA_INSTR = (
    "You are a vision assistant. Describe ONLY what you see in the image. "
    "Do not infer or hallucinate."
)

SYSTEM_LLAMA_INSTR = (
    "You are an expert assistant. Base your answer ONLY on the provided context "
    "and image descriptions. If the answer is not in the context, respond “I don't know.” "
    "Limit your response to 300 characters."
)

# ----------------------------------------
# Helper: Run Ollama locally via subprocess
# ----------------------------------------
def run_ollama(model_name, prompt):
    try:
        result = subprocess.run(
            ["ollama", "run", model_name],
            input=prompt.encode("utf-8"),
            capture_output=True,
            timeout=60
        )
        output = result.stdout.decode("utf-8").strip()
        return output
    except Exception as e:
        print(f"❌ Error running model '{model_name}': {e}")
        return ""

# ----------------------------------------
# Load prompts from Step 4
# ----------------------------------------
with open(PROMPTS_FILE, "r", encoding="utf-8") as pf:
    raw = pf.read().strip()

prompts = [p.strip() for p in raw.split("=" * 80) if p.strip()]

# ----------------------------------------
# Step 5: Generate answers using local LLaVA + LLaMA
# ----------------------------------------
answers = []

for idx, prompt in enumerate(prompts, start=1):
    print(f"\n🔍 Processing Prompt {idx}")

    lines = prompt.splitlines()
    question = lines[0].replace("Question:", "").strip()
    images_idx = next((i for i, l in enumerate(lines) if l.startswith("Images:")), len(lines))
    context = "\n".join(
        l for l in lines[1:images_idx]
        if not l.startswith("Context:")
    ).strip()
    images_line = lines[images_idx].replace("Images:", "").strip()
    image_paths = [p.strip() for p in images_line.split(",") if p.strip()]

    # ---- 1) Describe images via LLaVA ----
    image_descriptions = []
    for img_path in image_paths:
        llava_prompt = (
            f"{SYSTEM_LLAVA_INSTR}\n\n"
            f"Image:\n![]({os.path.abspath(img_path)})\n\n"
            "Description:"
        )
        desc = run_ollama(LLAVA_MODEL, llava_prompt)
        if desc:
            image_descriptions.append(desc)

    # ---- 2) Generate textual answer via LLaMA ----
    image_block = "\n".join(f"- {d}" for d in image_descriptions)
    llama_prompt = (
        f"System:\n{SYSTEM_LLAMA_INSTR}\n\n"
        f"Question: {question}\n\n"
        f"Context:\n{context}\n\n"
        f"ImageDescriptions:\n{image_block}\n\n"
        "Answer:"
    )
    answer = run_ollama(LLAMA_MODEL, llama_prompt)
    answer = answer.replace("\n", " ")[:MAX_ANSWER_CHARS]

    answers.append({
        "question": question,
        "answer": answer
    })
    print(f"➡️ Answer {idx}: {answer}")
```

```python
import os
import json
import requests
from json.decoder import JSONDecodeError
from requests import Response

# ----------------------------------------
# Configuration for Step 5
# ----------------------------------------
PROMPTS_FILE       = "prompts.txt"            # נוצר בשלב 4
OLLAMA_URL         = "http://localhost:11434/completions"
LLAVA_MODEL        = "llava-13b"              # החלף לשם המודל שמותקן אצלך
LLAMA_MODEL        = "llama-2-13b"
MAX_IMAGE_TOKENS   = 50                       # מספיק לתיאור קצר
MAX_ANSWER_TOKENS  = 100                      # ≈300 תווים בגוף תשובה

SYSTEM_LLAVA_INSTR = (
    "You are a vision assistant. Describe ONLY what you see in the image. "
    "Do not infer or hallucinate."
)

SYSTEM_LLAMA_INSTR = (
    "You are an expert assistant. Base your answer ONLY on the provided context "
    "and image descriptions. If the answer is not in the context, respond “I don't know.” "
    "Limit your response to 300 characters."
)

# ----------------------------------------
# Helper: Safe JSON loader
# ----------------------------------------
def safe_json(resp: Response) -> dict:
    text = resp.text.strip()
    try:
        return resp.json()
    except JSONDecodeError:
        pass
    # NDJSON / SSE fallback
    for line in text.splitlines():
        line = line.strip()
        if line.startswith("{") and line.endswith("}"):
            try:
                return json.loads(line)
            except JSONDecodeError:
                continue
    # Chop from first '{'
    idx = text.find("{")
    if idx != -1:
        try:
            return json.loads(text[idx:])
        except Exception:
            pass
    raise ValueError(f"Could not decode JSON from response:\n{text[:200]}")

# ----------------------------------------
# Load prompts created in Step 4
# ----------------------------------------
with open(PROMPTS_FILE, "r", encoding="utf-8") as pf:
    raw = pf.read().strip()

# Each prompt is separated by a line of 80 '=' characters
prompts = [p.strip() for p in raw.split("=" * 80) if p.strip()]

# ----------------------------------------
# Step 5: Generate answers
# ----------------------------------------
answers = []

for idx, prompt in enumerate(prompts, start=1):
    print(f"🔍 Processing Prompt {idx}")

    # Split into lines and parse
    lines = prompt.splitlines()
    question = lines[0].replace("Question:", "").strip()

    # Find where "Images:" begins
    images_idx = next((i for i, l in enumerate(lines) if l.startswith("Images:")), len(lines))
    # Context is everything between the header and the Images line
    context = "\n".join(
        l for l in lines[1:images_idx]
        if not l.startswith("Context:")
    ).strip()

    # Parse image paths
    images_line = lines[images_idx].replace("Images:", "").strip()
    image_paths = [p.strip() for p in images_line.split(",") if p.strip()]

    # ---- 1) Describe images via LLaVA ----
    image_descriptions = []
    for img_path in image_paths:
        llava_payload = {
            "model": LLAVA_MODEL,
            "prompt": (
                f"{SYSTEM_LLAVA_INSTR}\n\n"
                f"Image:\n![]({os.path.abspath(img_path)})\n\n"
                "Description:"
            ),
            "max_tokens": MAX_IMAGE_TOKENS
        }
        resp = requests.post(OLLAMA_URL, json=llava_payload)
        data = safe_json(resp)
        desc = data["choices"][0]["message"]["content"].strip()
        image_descriptions.append(desc)

    # ---- 2) Generate text answer via LLaMA ----
    image_block = "\n".join(f"- {d}" for d in image_descriptions)
    llama_payload = {
        "model": LLAMA_MODEL,
        "prompt": (
            f"System:\n{SYSTEM_LLAMA_INSTR}\n\n"
            f"Question: {question}\n\n"
            f"Context:\n{context}\n\n"
            f"ImageDescriptions:\n{image_block}\n\n"
            "Answer:"
        ),
        "max_tokens": MAX_ANSWER_TOKENS
    }
    resp = requests.post(OLLAMA_URL, json=llama_payload)
    data = safe_json(resp)
    answer = data["choices"][0]["message"]["content"].strip()

    # Enforce one-line, 300-character limit
    answer = answer.replace("\n", " ")[:300]

    answers.append({
        "question": question,
        "answer": answer
    })
    print(f"➡️ Answer {idx}: {answer}\n")

# 'answers' now holds the generated answers for each prompt
```

```python
#!/usr/bin/env python3
# coding: utf-8

"""
qa_patent_local.py

• עיבוד PDF באמצעות llava (ליצירת תיאורי תמונה דרך Ollama)  
• אינדוקס טקסט ב-Qdrant עם SentenceTransformer  
• שאילתות ותשובות עם llama3.2:1b דרך Ollama  
• תשובות עד 300 תווים, ללא הזיות – רק על סמך הקונטקסט שאוחזר  
"""
import re 
import os
import sys
import fitz
import numpy as np
import easyocr
import subprocess
import argparse
from PIL import Image
from sentence_transformers import SentenceTransformer
from qdrant_client import QdrantClient
from qdrant_client.http.models import VectorParams, PointStruct

# ----------------------------------------
# התאמה לסביבת Jupyter
# ----------------------------------------
if any("ipykernel" in k for k in sys.modules.keys()):
    sys.argv = [sys.argv[0]]

# ----------------------------------------
# ניתוח ארגומנטים
# ----------------------------------------
def parse_args():
    ap = argparse.ArgumentParser(description="Local Patent Q&A (PDF → Qdrant → Ollama)")
    
    # Input and output paths
    ap.add_argument("--pdf", default="patent.pdf", help="Path to the input PDF file")
    ap.add_argument("--images_dir", default="images", help="Directory to save extracted images")
    ap.add_argument("--questions", default="questions.txt", help="Path to the questions file")
    ap.add_argument("--answers", default="answers.txt", help="Path to save generated answers")

    # Model and collection configuration
    ap.add_argument("--collection", default="patent_chunks", help="Name of the Qdrant collection")
    ap.add_argument("--llava", default="llava", help="Ollama model name for image captioning")
    ap.add_argument("--llama", default="llama3.2:1b", help="Ollama model name for answer generation")
    ap.add_argument("--embed", default="all-MiniLM-L6-v2", help="SentenceTransformer model name")

    # Retrieval and filtering parameters
    ap.add_argument("--min_score", type=float, default=0.325, help="Minimum similarity score for context filtering")
    ap.add_argument("--top_k", type=int, default=4, help="Number of top context chunks to retrieve")
    ap.add_argument("--max_ctx_bytes", type=int, default=2000, help="Maximum byte size of context passed to LLM")

    return ap.parse_known_args()


ARGS, _ = parse_args()

# ----------------------------------------
# OCR וטעינת מודלים
# ----------------------------------------
ocr_reader = easyocr.Reader(["en"], gpu=False)

try:
    embedder = SentenceTransformer(ARGS.embed)
except Exception as e:
    print(f"⛔ שגיאה בטעינת embedder: {e}")
    sys.exit(1)

print("✅ מודל embedder נטען בהצלחה")

# ----------------------------------------
# עיבוד תמונה לתיאור עם Ollama
# ----------------------------------------
def caption_image(path):
    prompt = "Describe this image in one sentence."
    proc = subprocess.run(
        ["ollama", "run", ARGS.llava, "--image", path, "--prompt", prompt],
        capture_output=True, text=True
    )
    if proc.returncode != 0:
        return "⛔ תיאור נכשל"
    return proc.stdout.strip()

# ----------------------------------------
# פיצול טקסט ל־chunks
# ----------------------------------------
def chunk_text(text, max_chars=800):
    chunks = []
    i = 0
    while i < len(text):
        j = min(i + max_chars, len(text))
        if j < len(text):
            sep = max(text.rfind(".", i, j),
                      text.rfind("\n", i, j),
                      text.rfind(" ", i, j))
            if sep > i:
                j = sep + 1
        c = text[i:j].strip()
        if len(c) > 20:
            chunks.append(c)
        i = j
    return chunks

# ----------------------------------------
# תנאי הכנה
# ----------------------------------------
os.makedirs(ARGS.images_dir, exist_ok=True)

try:
    doc = fitz.open(ARGS.pdf)
except Exception as e:
    print(f"⛔ לא ניתן לפתוח PDF: {e}")
    sys.exit(1)

# ----------------------------------------
# אינדוקס ב-Qdrant
# ----------------------------------------
client = QdrantClient(":memory:")
points, texts_meta = [], []
dim = embedder.get_sentence_embedding_dimension()

client.create_collection(
    collection_name=ARGS.collection,
    vectors_config=VectorParams(size=dim, distance="Cosine")
)

for p in range(len(doc)):
    page = doc.load_page(p)
    page_num = p + 1

    # 1. טקסט מוטמע או OCR
    txt = page.get_text().strip()
    if not txt:
        pix = page.get_pixmap(dpi=300)
        arr = np.frombuffer(pix.samples, dtype=np.uint8)
        arr = arr.reshape(pix.height, pix.width, pix.n)
        ocr = ocr_reader.readtext(arr)
        txt = " ".join([l[1] for l in ocr]).strip()

    if txt:
        for chunk in chunk_text(txt):
            texts_meta.append({"type": "text", "page": page_num, "content": chunk})
            points.append(chunk)

    # 2. תמונות inline ותיאורי llava
    for idx, img_meta in enumerate(page.get_images(full=True), start=1):
        xref = img_meta[0]
        img_dict = doc.extract_image(xref)
        ext      = img_dict["ext"]
        fname    = f"page_{page_num:03d}_img_{idx}.{ext}"
        path     = os.path.join(ARGS.images_dir, fname)
        with open(path, "wb") as f:
            f.write(img_dict["image"])
        caption = caption_image(path)
        texts_meta.append({
            "type":    "image",
            "page":    page_num,
            "content": path,
            "caption": caption
        })
        points.append(caption)

# 3. וקטוריזציה ואחסון
vectors = embedder.encode(points, show_progress_bar=False)
for i, vec in enumerate(vectors):
    meta = texts_meta[i]
    client.upsert(
        collection_name=ARGS.collection,
        points=[
            PointStruct(
                id=i,
                vector=vec.tolist(),
                payload=meta
            )
        ]
    )

print("✅ כל הקטעים ואינדוקס Qdrant הושלמו")

# ----------------------------------------
# שלב 5: יצירת תשובות
# ----------------------------------------
def retrieve_context(query, top_k=ARGS.top_k):
    qv = embedder.encode([query])[0]
    hits = client.search(
        collection_name=ARGS.collection,
        query_vector=qv,
        limit=top_k
    )
    parts = []
    for hit in hits:
        payload = hit.payload
        if payload["type"] == "image":
            parts.append(f"[Image page {payload['page']}] {payload['caption']}")
        else:
            parts.append(payload["content"])
    return "\n".join(parts)

def generate_answer(question):
    ctx = retrieve_context(question)
    if not ctx.strip():
        return None

    prompt = (
        "Answer concisely using ONLY the context below. "
        "If the answer is not found in the context, reply: 'Not found in context.' "
        "Do not add ANYTHING else. "
        "Limit to 300 characters. Be direct and factual.\n\n"
        "Context:\n" + ctx + "\n\n"
        "Question: " + question + "\nAnswer:"
    )

    try:
        proc = subprocess.run(
            ["ollama", "run", ARGS.llama],
            input=prompt,
            capture_output=True,
            text=True
        )
    except Exception:
        return None

    if proc.returncode != 0:
        return None

    output = proc.stdout.strip()
    return output[:300] if output else None


# ----------------------------------------
# קריאת שאלות וכתיבת תשובות
# ----------------------------------------
try:
    with open(ARGS.questions, "r", encoding="utf-8") as f:
        questions = [line.strip() for line in f if line.strip()]
except Exception as e:
    print(f"⛔ שגיאה בטעינת {ARGS.questions}: {e}")
    sys.exit(1)

with open(ARGS.answers, "w", encoding="utf-8") as out:
    print("✅ מתחיל לענות על שאלות מהקובץ")
    for q in questions:
        print(f"שאלה: {q}")
        ans = generate_answer(q)
        if ans:
            print("תשובה:", ans)
            out.write(f"שאלה: {q}\nתשובה: {ans}\n\n")
        else:
            print("⛔ יצירת תשובה נכשלה")
            out.write(f"שאלה: {q}\nתשובה: ⛔ יצירת תשובה נכשלה\n\n")

print(f"✅ כל התשובות נשמרו בקובץ {ARGS.answers}")
```

```python
#!/usr/bin/env python3
# coding: utf-8

"""
qa_patent_local.py

• עיבוד PDF באמצעות llava (ליצירת תיאורי תמונה דרך Ollama)
• אינדוקס טקסט ב-Qdrant עם SentenceTransformer
• שאילתות ותשובות עם llama3.2:1b דרך Ollama
• תשובות עד 300 תווים, ללא הזיות – רק על סמך הקונטקסט שאוחזר
"""

import re
import os
import sys
import fitz
import numpy as np
import easyocr
import subprocess
import argparse
from PIL import Image
from sentence_transformers import SentenceTransformer
from qdrant_client import QdrantClient
from qdrant_client.http.models import VectorParams, PointStruct, Distance

# ----------------------------------------
# התאמה לסביבת Jupyter
# ----------------------------------------
if any("ipykernel" in k for k in sys.modules.keys()):
    sys.argv = [sys.argv[0]]

# ----------------------------------------
# ניתוח ארגומנטים
# ----------------------------------------
def parse_args():
    ap = argparse.ArgumentParser(description="Local Patent Q&A (PDF → Qdrant → Ollama)")
    
    # Input and output paths
    ap.add_argument("--pdf", default="patent.pdf", help="Path to the input PDF file")
    ap.add_argument("--images_dir", default="images", help="Directory to save extracted images")
    ap.add_argument("--questions", default="questions.txt", help="Path to the questions file")
    ap.add_argument("--answers", default="answers.txt", help="Path to save generated answers")

    # Model and collection configuration
    ap.add_argument("--collection", default="patent_chunks", help="Name of the Qdrant collection")
    ap.add_argument("--llava", default="llava", help="Ollama model name for image captioning")
    ap.add_argument("--llama", default="llama3.2:1b", help="Ollama model name for answer generation")
    ap.add_argument("--embed", default="all-MiniLM-L6-v2", help="SentenceTransformer model name")

    # Retrieval and filtering parameters
    ap.add_argument("--min_score", type=float, default=0.325, help="Minimum similarity score for context filtering")
    ap.add_argument("--top_k", type=int, default=4, help="Number of top context chunks to retrieve")
    ap.add_argument("--max_ctx_bytes", type=int, default=2000, help="Maximum byte size of context passed to LLM")

    return ap.parse_known_args()

ARGS, _ = parse_args()

# Global list to store all chunk metadata for easy access later
ALL_CHUNKS_META = []

# ----------------------------------------
# OCR וטעינת מודלים
# ----------------------------------------
ocr_reader = easyocr.Reader(["en"], gpu=False)

try:
    embedder = SentenceTransformer(ARGS.embed)
except Exception as e:
    print(f"⛔ שגיאה בטעינת embedder: {e}")
    sys.exit(1)

print("✅ מודל embedder נטען בהצלחה")

# ----------------------------------------
# עיבוד תמונה לתיאור עם Ollama
# ----------------------------------------
def caption_image(path):
    prompt = "Describe this image in one sentence."
    # Use explicit text=True to ensure proper decoding of output.
    proc = subprocess.run(
        ["ollama", "run", ARGS.llava, "--image", path, "--prompt", prompt],
        capture_output=True,
        text=True,
        encoding="utf-8" # Ensure UTF-8 encoding
    )
    if proc.returncode != 0:
        return "⛔ תיאור נכשל"
    return proc.stdout.strip()

# ----------------------------------------
# פיצול טקסט ל־chunks
# ----------------------------------------
def chunk_text(text, max_chars=800):
    chunks = []
    i = 0
    while i < len(text):
        j = min(i + max_chars, len(text))
        if j < len(text):
            sep = max(text.rfind(".", i, j),
                      text.rfind("\n", i, j),
                      text.rfind(" ", i, j))
            if sep > i:
                j = sep + 1
        c = text[i:j].strip()
        if len(c) > 20:
            chunks.append(c)
        i = j
    return chunks

# ----------------------------------------
# תנאי הכנה
# ----------------------------------------
os.makedirs(ARGS.images_dir, exist_ok=True)

try:
    doc = fitz.open(ARGS.pdf)
except Exception as e:
    print(f"⛔ לא ניתן לפתוח PDF: {e}")
    sys.exit(1)

# ----------------------------------------
# אינדוקס ב-Qdrant
# ----------------------------------------
client = QdrantClient(":memory:")
points = []
dim = embedder.get_sentence_embedding_dimension()

client.create_collection(
    collection_name=ARGS.collection,
    vectors_config=VectorParams(size=dim, distance=Distance.COSINE)
)

for p in range(len(doc)):
    page = doc.load_page(p)
    page_num = p + 1

    # 1. טקסט מוטמע או OCR
    txt = page.get_text().strip()
    if not txt:
        pix = page.get_pixmap(dpi=300)
        arr = np.frombuffer(pix.samples, dtype=np.uint8)
        arr = arr.reshape(pix.height, pix.width, pix.n)
        ocr = ocr_reader.readtext(arr)
        txt = " ".join([l[1] for l in ocr]).strip()

    if txt:
        for chunk in chunk_text(txt):
            meta = {"type": "text", "page": page_num, "content": chunk}
            ALL_CHUNKS_META.append(meta)
            points.append(chunk)

    # 2. תמונות inline ותיאורי llava
    for idx, img_meta in enumerate(page.get_images(full=True), start=1):
        xref = img_meta[0]
        img_dict = doc.extract_image(xref)
        ext      = img_dict["ext"]
        fname    = f"page_{page_num:03d}_img_{idx}.{ext}"
        path     = os.path.join(ARGS.images_dir, fname)
        with open(path, "wb") as f:
            f.write(img_dict["image"])
        caption = caption_image(path)
        meta = {
            "type":    "image",
            "page":    page_num,
            "content": path,
            "caption": caption
        }
        ALL_CHUNKS_META.append(meta)
        points.append(caption)

# 3. וקטוריזציה ואחסון
vectors = embedder.encode(points, show_progress_bar=False)
for i, vec in enumerate(vectors):
    meta = ALL_CHUNKS_META[i]
    client.upsert(
        collection_name=ARGS.collection,
        points=[
            PointStruct(
                id=i,
                vector=vec.tolist(),
                payload=meta
            )
        ]
    )

print("✅ כל הקטעים ואינדוקס Qdrant הושלמו")

# ----------------------------------------
# שלב 4: יצירת תשובות
# ----------------------------------------
def retrieve_context(query, top_k=ARGS.top_k):
    print("⏳ מחפש קונטקסט")
    query_vector = embedder.encode(query).tolist()
    
    hits = client.search(
        collection_name=ARGS.collection,
        query_vector=query_vector,
        limit=top_k,
        with_payload=True,
    )

    text_chunks = []
    image_paths = []
    
    # Store pages of text chunks to find nearby images
    text_pages = set()

    # 1. Get top-k relevant text chunks
    for hit in hits:
        payload = hit.payload
        if payload["type"] == "text":
            text_chunks.append(payload["content"])
            text_pages.add(payload["page"])
    
    # 2. Select up to 2 nearby image chunks
    for page in text_pages:
        for chunk in ALL_CHUNKS_META:
            if chunk["type"] == "image" and abs(chunk["page"] - page) <= 1:
                if chunk["content"] not in image_paths:
                    image_paths.append(chunk["content"])
                if len(image_paths) >= 2:
                    break
        if len(image_paths) >= 2:
            break

    full_context_text = "\n".join(text_chunks)
    # Enforce max byte size for context
    context_bytes = full_context_text.encode('utf-8')[:ARGS.max_ctx_bytes]
    limited_context_text = context_bytes.decode('utf-8', 'ignore')

    print("✅ קונטקסט נמצא בהצלחה")
    return limited_context_text, image_paths
def dynamic_top_k(question):
    if len(question.split()) > 15:
        return 6
    elif "diagram" in question.lower():
        return 8
    else:
        return 4

def generate_answer(question):
    ctx, images = retrieve_context(question, top_k=dynamic_top_k(question))

    
    if not ctx.strip():
        return None

    # Construct the base prompt.
    prompt = (
        "Answer concisely using ONLY the context below. "
        "If the answer is not found in the context, reply: 'Not found in context.' "
        "Do not add ANYTHING else. "
        "Limit to 300 characters. Be direct and factual.\n\n"
        "Context:\n" + ctx + "\n\n"
        "Question: " + question + "\nAnswer:"
    )

    # Use the Llama model for text generation and pass the entire prompt as input.
    # The image descriptions are already included in the `ctx` variable.
    try:
        proc = subprocess.run(
            ["ollama", "run", ARGS.llama],
            input=prompt,
            capture_output=True,
            text=True,
            encoding="utf-8"
        )
    except Exception as e:
        print(f"⛔ שגיאה בהפעלת Ollama: {e}")
        return None

    if proc.returncode != 0:
        print(f"⛔ Ollama החזיר קוד שגיאה: {proc.returncode}")
        print(f"stdout: {proc.stdout}")
        print(f"stderr: {proc.stderr}")
        return None

    output = proc.stdout.strip()
    return output[:300] if output else None

# ----------------------------------------
# קריאת שאלות וכתיבת תשובות
# ----------------------------------------
try:
    with open(ARGS.questions, "r", encoding="utf-8") as f:
        questions = [line.strip() for line in f if line.strip()]
except Exception as e:
    print(f"⛔ שגיאה בטעינת {ARGS.questions}: {e}")
    sys.exit(1)

with open(ARGS.answers, "w", encoding="utf-8") as out:
    print("✅ מתחיל לענות על שאלות מהקובץ")
    for q_idx, q in enumerate(questions):
        print(f"שאלה: {q}")
        ans = generate_answer(q)
        if ans:
            print("תשובה:", ans)
            out.write(f"Question {q_idx + 1}: {q}\n{ans}\n\n")
        else:
            print("⛔ יצירת תשובה נכשלה")
            out.write(f"Question {q_idx + 1}: {q}\nNot found in context.\n\n")

print(f"✅ כל התשובות נשמרו בקובץ {ARGS.answers}")
```
